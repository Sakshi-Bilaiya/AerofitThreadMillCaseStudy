## Customer Profiling – Business Recommendations

Based on the customer profiles, product distributions, and relationships between **Age, Income, Usage, Fitness, and Miles**, the following recommendations can be made.

### 1. KP281 — Target general fitness customers

**Who to target:**

* Customers with moderate fitness levels
* Customers expecting moderate treadmill usage
* Customers across a broad range of ages and incomes
* Customers who are new or less intensive treadmill users

**What to do:**

* Position KP281 as a practical option for regular, everyday exercise.
* Promote it to customers who want a treadmill for moderate workouts.
* Use simple messaging around convenience and maintaining a regular fitness routine.

---

### 2. KP481 — Target regular fitness users

**Who to target:**

* Customers with moderate-to-good fitness levels
* Customers planning regular treadmill use
* Customers whose requirements fall between basic and intensive usage

**What to do:**

* Position KP481 as an option for customers who want more than basic treadmill use.
* Highlight its suitability for customers following a consistent workout routine.
* Target customers who are upgrading their fitness routine but may not need the most intensive option.

---

### 3. KP781 — Target high-activity customers

**Who to target:**

* Customers with higher fitness levels
* Customers planning frequent treadmill use
* Customers expecting to cover more miles
* Customers with higher incomes

**What to do:**

* Position KP781 toward customers looking for a more intensive workout experience.
* Focus marketing on frequent users and fitness-focused customers.
* Highlight features that matter to customers who expect heavier or more frequent use.
* Consider targeted promotions for higher-income fitness customers.

---

## 4. Actionable marketing recommendations

| Product   | Target customer                               | Business action                                    |
| --------- | --------------------------------------------- | -------------------------------------------------- |
| **KP281** | Moderate users and general fitness customers  | Promote as an everyday fitness option              |
| **KP481** | Regular users with moderate-to-good fitness   | Position as an upgrade for consistent workouts     |
| **KP781** | High-fitness, frequent and high-mileage users | Target fitness-focused and higher-income customers |

### 5. Use customer characteristics together

Rather than targeting customers based on only one characteristic, combine several signals.

For example, a customer with:

* High fitness
* Frequent expected usage
* High expected miles
* Higher income

matches the observed profile associated more closely with **KP781**.

Similarly, customers with moderate usage and fitness can be considered for **KP281 or KP481**, depending on their requirements.

---

## Final Business Takeaway

> **KP281 should be marketed broadly to customers looking for regular, moderate exercise. KP481 can be positioned toward customers with a more consistent fitness routine who want an intermediate option. KP781 should be targeted toward frequent, high-activity customers, particularly those with higher fitness levels, higher expected mileage, and higher incomes.**

These are **data-based targeting suggestions from this dataset**, not guarantees that a particular customer will purchase a particular product.
